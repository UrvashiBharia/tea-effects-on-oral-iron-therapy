```r
set.seed(42)
n <- 200

## predictors
tea_levels <- c(1.5, 1.65, 2.5, 5, 7.5, 10, 15, 20)
tea_g <- sample(tea_levels, n, replace = TRUE,
                prob = c(2, 5, 23, 37, 5, 35, 15, 12))
therapy_months <- sample(1:6, n, replace = TRUE, prob = c(8, 10, 15, 15, 15, 37))

age      <- pmin(pmax(round(rnorm(n, mean = 32, sd = 10)), 15), 49)   #15-49
hb_start <- pmax(round(rnorm(n, mean = 10.2, sd = 1.2), 1), 8)         #floor at 8 g/dl
```

```r
#variables
true_intercept <- 0.50
true_beta_tea  <- -0.020
true_beta_dur  <-  0.10
noise_sd       <-  0.50

hb_change <- true_intercept +
             true_beta_tea * tea_g +
             true_beta_dur * therapy_months +
             rnorm(n, mean = 0, sd = noise_sd)
hb_change <- round(hb_change, 2)
hb_end    <- round(hb_start + hb_change, 2)
locality <- sample(c("Dharavi","Matunga","Kurla","Ghatkopar","Chembur","Sion"),
                    n, replace = TRUE)
freq <- round(tea_g / 5, 1)   #cups/day

compliant_p <- 0.80
oit <- ifelse(
  rbinom(n, 1, compliant_p) == 1,
  sample(c("yes- breakfast","yes- lunch","yes- dinner","yes"), n, replace = TRUE),
  sample(c("no","rarely","rarely- breakfast"), n, replace = TRUE)
)
black_stool  <- rbinom(n, 1, 0.30)
constipation <- rbinom(n, 1, 0.20)
```

```r
#summary
sim <- data.frame(
  id = 1:n, hb_change, tea_g, therapy_months, hb_start, hb_end,
  age, locality, freq, oit, black_stool, constipation
)
write.csv(sim, "practice_tea_data.csv", row.names = FALSE)
```
